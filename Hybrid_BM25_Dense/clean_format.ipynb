# Format Hybrid Union Results

將 `outputs/hybrid_union_results.json` 轉成和 `query1_result.jsonl` 相同的 `jsonl` 格式，並依 `rank` 排序。

```python
from __future__ import annotations

import json
from pathlib import Path

INPUT_PATH = Path("/workplace/dxlin/Homework/IR/Hybrid_BM25_Dense/outputs/hybrid_union_results.json")
OUTPUT_PATH = Path("/workplace/dxlin/Homework/IR/Hybrid_BM25_Dense/query2_result.jsonl")
TOP_K = 30


def build_record(raw_item: dict, item_id: int) -> dict:
    sorted_answers = sorted(raw_item.get("results", []), key=lambda answer: answer.get("rank", float("inf")))[:TOP_K]

    return {
        "id": item_id,
        "question": raw_item.get("query", ""),
        "answers": [
            {
                "id": answer.get("doc_id", ""),
                "text": answer.get("answer", ""),
            }
            for answer in sorted_answers
        ],
    }


raw_data = json.loads(INPUT_PATH.read_text(encoding="utf-8"))
formatted_records = [build_record(item, idx) for idx, item in enumerate(raw_data, start=1)]

with OUTPUT_PATH.open("w", encoding="utf-8") as output_file:
    for record in formatted_records:
        output_file.write(json.dumps(record, ensure_ascii=False) + "\n")

print(f"Wrote {len(formatted_records)} records to {OUTPUT_PATH}")
formatted_records[0]
```

把JSONL轉成JSON，比較好讀，但正式提交要提交JSONL。

```python
from __future__ import annotations

import json
from pathlib import Path

JSONL_INPUT_PATH = Path("/workplace/dxlin/Homework/IR/Hybrid_BM25_Dense/query2_result.jsonl")
JSON_OUTPUT_PATH = Path("/workplace/dxlin/Homework/IR/Hybrid_BM25_Dense/query2_result.json")

records = [
    json.loads(line)
    for line in JSONL_INPUT_PATH.read_text(encoding="utf-8").splitlines()
    if line.strip()
]

JSON_OUTPUT_PATH.write_text(
    json.dumps(records, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

print(f"Wrote {len(records)} records to {JSON_OUTPUT_PATH}")
records[0]
```

把 `hybrid_union_query1_results.json` 轉成 `exp/scripts/evaluate_dense_rerank.py` 可直接評估的格式。

```python
from __future__ import annotations

import json
from pathlib import Path

HYBRID_INPUT_PATH = Path("/workplace/dxlin/Homework/IR/Hybrid_BM25_Dense/outputs/hybrid_union_query1_results.json")
EVAL_OUTPUT_PATH = Path("/workplace/dxlin/Homework/IR/exp/outputs/hybrid_union_query1_results_for_eval.json")


def build_eval_record(raw_item: dict) -> dict:
    sorted_results = sorted(
        raw_item.get("results", []),
        key=lambda answer: answer.get("rank", float("inf")),
    )

    return {
        "query_id": raw_item.get("query_id", ""),
        "query": raw_item.get("query", ""),
        "rerank_results": [
            {
                "rank": answer.get("rank"),
                "doc_id": answer.get("doc_id", ""),
                "question": answer.get("question", ""),
                "answer": answer.get("answer", ""),
                "rerank_score": answer.get("rerank_score"),
            }
            for answer in sorted_results
        ],
    }


hybrid_results = json.loads(HYBRID_INPUT_PATH.read_text(encoding="utf-8"))
eval_records = [build_eval_record(item) for item in hybrid_results]

EVAL_OUTPUT_PATH.write_text(
    json.dumps(eval_records, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

print(f"Wrote {len(eval_records)} records to {EVAL_OUTPUT_PATH}")
eval_records[0]
```
